## Models

### Agglomerative Clustering
- Does not assume spherical clusters
- Works well with hierarchical data

```python
# AgglomerativeClustering model with # clusters = 5 (5 cancer types), ward linkage, and euclidean distance
agg_model = AgglomerativeClustering(n_clusters=5, linkage='ward', metric='euclidean')
# fit the model
agg_predictions = agg_model.fit_predict(data)
# print out results
print("Cluster Labels:")
print(agg_predictions)
```

### KMeans Clustering with PCA Reduced Data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(data)

pca = PCA().fit(X_scaled)
explained = np.cumsum(pca.explained_variance_ratio_)

plt.figure(figsize=(8,4))
plt.plot(explained, marker = 'o')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA Explained Variance')
plt.grid(True)
plt.tight_layout()
# save fig optional
#plt.savefig('results/pca_explained_variance.jpg')
plt.show()
```

**Given a fixed number of clusters (k = 5), how does PCA dimensionality affect cluster quality?**
- Silhouette scores measure the separation between clusters, with higher scores representing better separation.

```python
candidate_dims = [5, 10, 50, 100, 250, 500]
sil_scores = []

for k in candidate_dims:
    pca = PCA(n_components=k)
    X_pca = pca.fit_transform(X_scaled)
    
    model = KMeans(n_clusters=5, random_state=42)
    preds = model.fit_predict(X_pca)
    
    sil_score = silhouette_score(X_pca, preds)
    sil_scores.append(sil_score)

plt.figsize=(6, 4)
plt.plot(candidate_dims, sil_scores, marker='o', color='cornflowerblue')
plt.ylabel("Silhouette Score")
plt.xlabel("PCA Components")
plt.title("Optimal PCA Dimensions for KMeans")

plt.tight_layout()
# save fig optional
#plt.savefig('results/optimal_pca_dims_kmeans.jpg') 
plt.show()
```

```python
# K-Means model with # clusters = 5 (5 cancer types), PCA Reduced Data with 10 components
pca = PCA(n_components=10)
X_pca = pca.fit_transform(X_scaled)
km_model = KMeans(n_clusters=5, random_state=42)
# fit the model
km_predictions = km_model.fit_predict(X_pca)
# print out results
print("Cluster Labels:")
print(km_predictions)
```

```python
import itertools

def attach_labels(labels,ypred,n=5):
    """
    Function to attach labels to clusters based on best accuracy
    
    labels: labels dataframe object
    ypred: clustering model predictions
    
    Returns permuted label order and accuracy. 
    """
    # actual labels and predicted labels
    actual_labels = labels['Class'].values  
    label_options = ['BRCA', 'KIRC', 'COAD', 'LUAD', 'PRAD']

    # initialize variables to store results
    acc = 0.0
    best_label_order = None

    # all possible label orderings
    label_permutations = itertools.permutations(label_options)

    for label_order in label_permutations:
        # mapping from numeric cluster labels to the current label permutation
        label_mapping = {i: label_order[i] for i in range(n)}

        # map the cluster labels to actual labels using the current mapping
        mapped_labels = np.array([label_mapping[label] for label in ypred])

        # accuracy of the current label permutation
        accuracy = accuracy_score(actual_labels, mapped_labels)

        # determine best accuracy and label order
        if accuracy > acc:
            acc = accuracy
            best_label_order = label_order
            
    # mapping from numeric labels to best permuted labels
    best_label_mapping = {i: best_label_order[i] for i in range(n)}
    # get the best label assignments
    best_labels_with_assignment = [best_label_mapping[label] for label in ypred]

    return best_label_mapping, acc
    
```

```python
agg_labelorder, agg_acc = attach_labels(labels, agg_model.labels_)
print(agg_labelorder, agg_acc)
```

```python
# Mapped predictions
mapped_predictions = [agg_labelorder[label] for label in agg_model.labels_]
# Create confusion matrix
y_true = labels['Class'].values
y_pred = mapped_predictions

cm = confusion_matrix(y_true, y_pred, labels=['BRCA', 'KIRC', 'COAD', 'LUAD', 'PRAD'])

# Display confusion matrix
plt.figure(figsize=(4,4))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['BRCA', 'KIRC', 'COAD', 'LUAD', 'PRAD'])
disp.plot(cmap='Blues', xticks_rotation=45)
plt.title("Confusion Matrix: Agglomerative Clustering")
plt.tight_layout()
# save matrix, optional
#plt.savefig('results/confusion_matrix_agg_clustering.jpg')
plt.show()
```

## Heatmap of variable genes

The dendrogram above the heatmap represents hierarchical clustering of genes based on similarity in their expression profiles across samples. Genes that cluster together exhibit similar patterns of variation, suggesting potential co-expression or shared biological regulation. The data will be z-score normalized using StandardScaler and gene expression is in standard deviation units: 

- 0 (white blocks) indicates average expression of that gene across samples.
- Positive values (red blocks) indicate higher-than-average expression.
- Negative values (blue blocks) indicate lower-than-average expression.

```python
# Compute variance per gene and select number of genes to visualize
gene_variance = data.var(axis=0)
top_n = 25
top_genes = gene_variance.sort_values(ascending=False).head(top_n).index
```

```python
# subset the data by top genes based on variance
data_hvg = data[top_genes]

# scale the data, z-score normalization per gene
scaler = StandardScaler()
data_hvg_scaled = scaler.fit_transform(data_hvg)

data_hvg_scaled = pd.DataFrame(
    data_hvg_scaled,
    index=data.index,
    columns=top_genes
)
```

```python
# Group the samples by cancer type
data_hvg_scaled['Class'] = labels['Class'].values
data_hvg_scaled = data_hvg_scaled.sort_values('Class')
```

```python
# Drop label for plotting
heatmap_data = data_hvg_scaled.drop(columns='Class')
```

```python
sns.clustermap(
    heatmap_data,
    cmap='vlag',
    row_cluster=False,
    col_cluster=True,
    figsize=(20, 12)
)

plt.tight_layout()
plt.show()
```

```python
# map cancer types to colors
cancer_palette = {
    'BRCA': 'tab:blue',
    'KIRC': 'tab:orange',
    'COAD': 'tab:green',
    'LUAD': 'tab:red',
    'PRAD': 'tab:purple'
}

legend_handles = [
    mpatches.Patch(color=color, label=label)
    for label, color in cancer_palette.items()
]

row_colors = labels.map(cancer_palette)

cluster_plot= sns.clustermap(
    heatmap_data,
    row_colors=row_colors,
    row_cluster=False,
    cmap='vlag',
    figsize=(10, 12)
)

# add legend
cluster_plot.ax_heatmap.legend(
    handles=legend_handles,
    title="Cancer Types",
    loc="upper right",
    bbox_to_anchor=(-0.05, 0.95),
    borderaxespad=0
)

plt.savefig('results/heatmap_most_varaible_genes_by_cancer_type')
plt.show()
```
